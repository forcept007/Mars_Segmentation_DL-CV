```python
import os
import matplotlib.pyplot as plt
from PIL import Image
```

```python
import os
print(os.getcwd())
```

```python
easy_path = 'S5Mars_data/images/easy'
hard_path = 'S5Mars_data/images/hard'
easy_files = sorted(os.listdir(easy_path))[:10]
hard_files = sorted(os.listdir(hard_path))[:10]
```

```python
fig, axes = plt.subplots(2, 5, figsize=(20, 8))

for ax, name in zip(axes.flat, easy_files):
    img_path = os.path.join(easy_path, name)
    img = Image.open(img_path)
    ax.imshow(img)
    ax.set_title(name, fontsize=8)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 5, figsize=(20, 8))

for ax, name in zip(axes.flat, hard_files):
    img_path = os.path.join(hard_path, name)
    img = Image.open(img_path)
    ax.imshow(img)
    ax.set_title(name, fontsize=8)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
